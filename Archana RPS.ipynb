```python
import sys
player1 = input("What's your name?")
player2 = input("And your name?")
player1_answer = input("%s, do yo want to choose rock, paper or scissors?" % player1)
player2_answer = input("%s, do you want to choose rock, paper or scissors?" % player2)
def compare(p1, p2):
    if p1 == p2:
        return("It's a tie!")
    elif p1 == 'rock':
        if p2 == 'scissors':
            return("Congratulations Rock! You won this round")
        else:
            return("Congratulations Paper! You won this round.")
    elif p1  == 'scissors':
        if p2 == 'paper':
            return("Congratulations Scissors! You won this round")
        else:
            return("Congratulations Rock! You won this round")
    elif p1  == 'paper':
        if p2 == 'rock':
            return("Congratulations Paper! You won this round.")
        else:
            return("Congratulations Scissors! You won this round")
    else:
        return("Invalid input! You have not entered rock, paper or scissors, try again.")
        sys.exit()

print(compare(player1_answer, player2_answer))
```
