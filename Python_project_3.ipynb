Cleaning Data

```python
drop_cols = ['X_COORD_CD','Y_COORD_CD', 'Latitude', 'Longitude', 'Location']
df.drop(drop_cols, inplace = True, axis=1)
df.head()
```

```python
import numpy as np
df = df.replace(['(null)'], np.nan)
df = df.dropna()
df.head()
```

Groping Data to show Total # offence by Boro


```python
print(df.groupby(['ARREST_BORO'])['OFNS_DESC'].count())
```

Visualization of Total # offence by Boro

```python
df.groupby(['ARREST_BORO'])['OFNS_DESC'].count().plot(kind='bar')
```

Groping Data to show Total # offence by Sex

```python
print(df.groupby(['PERP_SEX'])['OFNS_DESC'].count())
```

Visualization of Total # offence by Sex

```python
df.groupby(['PERP_SEX'])['OFNS_DESC'].count().plot(kind='bar')
```

Groping Data to show Total # offence by Boro as well as Sex

```python
df.groupby(['ARREST_BORO', 'PERP_SEX'])['OFNS_DESC'].count()
```

Descriptive Statistics

```python
print(df.groupby(['ARREST_BORO', 'PERP_SEX'])['OFNS_DESC'].count().describe())
```

```python
grouped_data = df.groupby(['ARREST_BORO', 'PERP_SEX'])['OFNS_DESC'].count()
plot_df = grouped_data.reset_index(name='COUNT')
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

Visualization Data to show Total # offence by Boro as well as Sex

```python
sns.catplot(data=plot_df, kind='bar', x='ARREST_BORO', y='COUNT', hue='PERP_SEX')
```

The data indentifies that there are more Male (M) arrest than female (F) in each Borough. However, Brooklyn (K) borough have more arrest compare to other boroughs. With that i find Statenisland (S) is safest borough to live.
