```python
cols = ['Mean Score', 'Percent Scoring Below 65', 'Percent Scoring 65 or Above', 'Percent Scoring 80 or Above']
```

```python
df2 = df2.dropna(subset=cols)
```

```python
last4 = df2.columns[-4:]
```

```python
df2 = df2.dropna(subset=last4)
```

```python
df2.tail()
```

```python
df2_compare = df2[df2["Year"].isin([2015,2016])]
```

```python
df2_compare["Mean Score"] = pd.to_numeric(df2_compare['Mean Score'], errors='coerce')
```

```python
print(df2_compare.groupby("Year")['Mean Score'].mean())
```

```python
df2_compare.groupby("Year")['Mean Score'].mean().plot(kind='bar')
```

I chose to compare Living Environment scores across borough of manhattan schools for 2015 vs 2016. The data shows that the mean score was little better in 2016 vs 2015. With this i conclude that the board of education worked to improve their test scores from 2015 to 2016 but not significant diffrence observed however, their average was above 70.
